```python
from vectordb import LSAEmbedder, VectorDB
```

```python
texts = [
    "Quarterly revenue grew 12% driven by loan interest income",
    "Revenue growth slowed as interest income declined",
    "Credit risk exposure rose due to non-performing loans",
    "Risk provisions increased for the mining sector",
    "Operating expenses fell after branch consolidation",
    "Branch closures reduced operating cost",
]

metadatas = [{"division": d} for d in ["finance", "finance", "risk", "risk", "ops", "ops"]]
```

```python
embedder = LSAEmbedder(dim=16, min_df=1)
db = VectorDB(embedder.output_dim if embedder.fit(texts) else 0)
db.upsert(embedder.embed(texts), texts, metadatas)
```

```python
for query, where in [("revenue growth", None), ("loan problems", {"division": "risk"}), ("revnue", None)]:
    print(f"\n{query!r} {where}")
    for r in db.search(embedder.embed(query), k=2, where=where):
        print(f"  {r.score:.3f}  [{r.metadata['division']}] {r.text}")
```
